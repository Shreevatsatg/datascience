```python
import json
from pathlib import Path
import pandas as pd

# 1. Define the data directory path using a raw string literal
data_dir = Path(r"D:\coding\datascience\Class5_Slide and EL\data")

# 2. Open and load the nested students.json file safely
with open(data_dir / "students.json") as f:
    records = json.load(f)

# 3. Use pd.json_normalize to flatten the loaded JSON data
flat = pd.json_normalize(records)

# 4. Print the list of columns available in the flattened DataFrame
print("Flattened Columns:", flat.columns.tolist())

# 5. Explode a column (replace 'skills_column_name' with the actual column you want to explode)
# 'flat.explode' requires a column name target to unpack nested lists into rows
# skills = flat.explode('skills_column_name')

# 6. Read a JSON Lines (.jsonl) file into a DataFrame
# Note: 'lines=True' must be capitalised, and 'dtype' forces specific columns to strings
logs = pd.read_json(
    data_dir / "students.jsonl", 
    lines=True, 
    dtype={"srn": "string"}
)

# Preview the logs DataFrame
print(logs.head())
```
